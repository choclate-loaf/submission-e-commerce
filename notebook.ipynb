## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

```python
orders_df['order_purchase_timestamp'] = pd.to_datetime(
    orders_df['order_purchase_timestamp'],
    errors='coerce'
)
```

```python
# buat kolom bulan
orders_df['month'] = orders_df['order_purchase_timestamp'].dt.to_period('M')

# hitung jumlah order
monthly_orders = orders_df.groupby('month')['order_id'].count()

monthly_orders.head()
```

- Jumlah order pada tahun 2016 masih sangat rendah, dengan hanya 4 order pada September dan 1 order pada Desember
- Terjadi peningkatan signifikan pada tahun 2017, khususnya pada bulan Januari dan Februari
- Jumlah order meningkat dari 324 pada Oktober 2016 menjadi 1780 pada Februari 2017
- Peningkatan ini menunjukkan adanya pertumbuhan aktivitas transaksi yang cukup pesat pada awal tahun 2017

##EDA 2: Tren Revenue Bulanan

```python
# gabungkan orders + payments
df = orders_df.merge(payments_df, on='order_id')

# hitung revenue bulanan
monthly_revenue = df.groupby('month')['payment_value'].sum()

monthly_revenue.head()
```

- Revenue pada awal periode (2016) masih relatif rendah dan belum stabil
- Terjadi peningkatan signifikan pada awal tahun 2017, khususnya pada bulan Januari dan Februari
- Lonjakan revenue pada tahun 2017 menunjukkan adanya peningkatan aktivitas transaksi pelanggan
- Fluktuasi revenue antar bulan mengindikasikan adanya pengaruh faktor musiman atau promosi tertentu

##EDA 3: RFM Analysis

```python
import datetime as dt

current_date = df['order_purchase_timestamp'].max()

rfm = df.groupby('customer_id').agg({
    'order_purchase_timestamp': lambda x: (current_date - x.max()).days,
    'order_id': 'count',
    'payment_value': 'sum'
})

rfm.columns = ['Recency', 'Frequency', 'Monetary']
rfm.head()
```

- Sebagian besar pelanggan memiliki nilai Frequency = 1, yang menunjukkan bahwa pelanggan cenderung hanya melakukan satu kali transaksi
- Nilai Recency yang cukup besar pada beberapa pelanggan menunjukkan bahwa mereka sudah lama tidak melakukan pembelian
- Nilai Monetary bervariasi, yang menunjukkan adanya perbedaan kontribusi setiap pelanggan terhadap revenue
- Pola ini menunjukkan bahwa mayoritas pelanggan belum menjadi pelanggan loyal

## Visualization & Explanatory Analysis

### Pertanyaan 1:
Bagaimana tren jumlah order dan total revenue bulanan selama periode 2017–2018 serta kapan terjadi puncak penjualan tertinggi?

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,5))

# Plot order
monthly_orders.plot(marker='o', label='Jumlah Order')

# Plot revenue
monthly_revenue.plot(marker='o', label='Revenue')

plt.title("Tren Order dan Revenue Bulanan")
plt.xlabel("Bulan")
plt.ylabel("Jumlah")
plt.legend()
plt.grid()
plt.show()
```

**Insight:** (Opsional)

* **Tren Naik**: Revenue menunjukkan peningkatan yang cukup konsisten dari awal periode hingga pertengahan–akhir 2017, menandakan pertumbuhan bisnis yang stabil.
* **Puncak Revenue**: Terjadi lonjakan signifikan pada sekitar Oktober–November 2017, yang menjadi titik tertinggi sepanjang periode pengamatan.
* **Fluktuasi**: Setelah mencapai puncak, revenue mengalami naik-turun di awal 2018 namun tetap berada di level tinggi (relatif stabil).
* **Jumlah Order**: Jumlah order cenderung stabil dan tidak mengalami kenaikan drastis seperti revenue.
* **Nilai Transaksi**: Perbedaan tren antara order dan revenue mengindikasikan adanya peningkatan rata-rata nilai transaksi (Average Order Value).
* **Akhir Periode**: Terjadi penurunan drastis hingga nol di bulan terakhir, kemungkinan disebabkan oleh data yang belum lengkap atau kendala operasional.

### Pertanyaan 2:
Bagaimana segmentasi pelanggan berdasarkan RFM (Recency, Frequency, Monetary) dan bagaimana kontribusi masing-masing segmen terhadap revenue?

```python
import seaborn as sns

plt.figure(figsize=(8,5))
sns.histplot(rfm['Frequency'], bins=10)
plt.title("Distribusi Frequency Pelanggan")
plt.xlabel("Jumlah Transaksi")
plt.ylabel("Jumlah Pelanggan")
plt.show()
```

**Insight:** (Opsional)
* **Mayoritas Pelanggan Frekuensi Rendah**: Sebagian besar pelanggan hanya melakukan 1–3 transaksi, terlihat dari distribusi yang menumpuk di sisi kiri.
* **Distribusi Tidak Merata (Right-Skewed)**: Grafik menunjukkan distribusi miring ke kanan, artinya hanya sedikit pelanggan dengan frekuensi transaksi tinggi.
* **Pelanggan Loyal Sedikit**: Hanya sebagian kecil pelanggan yang melakukan transaksi berulang dalam jumlah banyak (di atas 5–10 kali).
* **Potensi Retensi Rendah**: Banyaknya pelanggan dengan transaksi rendah mengindikasikan tingkat retensi masih bisa ditingkatkan.
* **Peluang Upselling & Engagement**: Perlu strategi untuk meningkatkan frekuensi pembelian, seperti promo, loyalty program, atau rekomendasi produk.
* **Segmentasi Pelanggan**: Data ini cocok digunakan untuk segmentasi (misalnya: pelanggan baru, aktif, dan loyal) guna strategi marketing yang lebih tepat sasaran.
